```python
# input and output
x = data.iloc[:,:1]
y = data.iloc[:,-1]
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test = train_test_split(x,y ,test_size = 0.2,random_state = 2)
```

```python
# apply the algorithms
from sklearn.linear_model import LinearRegression

obj = LinearRegression()
# .fit works => to train the data now look below
obj.fit(x_train,y_train) # => model are learning here.

```

```python
# import pickle
# pickle.dump(obj,open('model.pkl','wb'))
```

```python
# for the prediciton
obj.predict(x_test.iloc[1].values.reshape(1,1))
```

```python
# proper create.
plt.scatter(data['cgpa'],data['package'])
plt.plot(x_train,obj.predict(x_train),color = 'red')
plt.ylabel('package')
plt.xlabel('cgpa')
plt.title(" Package in per/lpa")
plt.show()
```
