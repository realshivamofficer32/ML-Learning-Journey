# Model training


```python
# take the input features do separate and output features
X = df.iloc[:,0:8]
y = df.iloc[:,-1]
```

```python
# check the shape of input and ouput features
```

```python
X.shape
```

```python
y.shape
```

```python
# divide the data into four parts so that we can predict after the a new data
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test = train_test_split(X,y,test_size = 0.2,random_state = 4)
```

```python
# apply the linear regression here
from sklearn.linear_model import LinearRegression
# make the object to the class of linearRegression
model = LinearRegression()
# train the model
model.fit(X_train,y_train)
```

```python
# predict the output
y_model = model.predict(X_test)
```

```python
X_test.head(5)
```

```python
y_test.head(5)
```

```python
X_test.iloc[0].values
```

```python
X_test.head(1)
```

```python
# check the accuracy of the model
from sklearn.metrics import mean_absolute_error,mean_squared_error,r2_score

# find the mean_absolute error
mae = mean_absolute_error(y_test,y_model)
# find the mean_squared_error
mse = mean_squared_error(y_test,y_model)

# root mean squared value =>

rmse = np.sqrt(mean_squared_error(y_test,y_model))


# find the score of model
r_score = r2_score(y_test,y_model)

# ====== > print all the values
print(" MAE : " ,mae)
print(" MSE : ",mse)
print(" R2_Score : ",r_score)
print(f" this is a RMSE(root mean squared error : {rmse:.4f}")
```

```python
y_test
```

```python
y_test.max()
```

```python
# find the error
residuals = y_test - y_model

```

```python
# plot the graph of residuls with actual number or value
plt.figure(figsize=(8,5))
sns.histplot(residuals, bins=30, kde=True, color='orange')
plt.axvline(x=0, color='green', linestyle='--', linewidth=3)
plt.title('Residuals (गलतियों) का वितरण', fontsize=14)
plt.xlabel('Residual (असली कीमत - हमारी भविष्यवाणी)', fontsize=12)
plt.ylabel('Frequency (कितनी बार आई ये गलती)', fontsize=12)
plt.grid(True, alpha=0.3)
plt.show()
```

```python
# - Graph of Residuals vs Predicted

plt.figure(figsize=(8,5))
# x - axes make the prediction and y -axes residuals or error show
plt.scatter(y_model, residuals, alpha=0.5, color='purple')
# draw a line in y axis
plt.axhline(y=0, color='red', linestyle='--', linewidth=2)
# give the title
plt.title('Residuals vs Predicted ', fontsize=14)
# X axes name provide here.
plt.xlabel('Predicted Price ', fontsize=12)
# y axes name proive here
plt.ylabel('Residuals ', fontsize=12)
# show grid
plt.grid(True, alpha=0.3)
plt.show()
```

```python
# Final Graph: Predicted vs Actual Scatter
plt.figure(figsize=(8,6))
plt.scatter(y_test, y_model, alpha=0.5, color='green')
plt.plot([y_test.min(), y_test.max()], [y_test.min(), y_test.max()], color = 'red', linewidth = 3, label='Perfect Line')
plt.xlabel('Actual Price ')
plt.ylabel('Predicted Price ')
plt.title('Actual vs Predicted House Price')
plt.legend()
plt.grid(True)
plt.show()
```
